# ***3. Clean Dataset: Check for duplicate records, etc...***

***A. Lets check for duplicate records***

```python
#Let's check for duplicate records
print("Number of exact duplicate records: ",df_spotify.duplicated().sum())
#Let's check for duplicate track_ids
print("Number of exact duplicate track_ids records: ",df_spotify["track_id"].duplicated().sum())
#Now let's check for duplicate track_ids and sort by artist name to see if we can find any clues as to why we see these duplicates
dup_trackName = df_spotify[df_spotify["track_id"].duplicated(keep=False)].sort_values(by="artist_name")
display(dup_trackName)
```

***A. Conclusion of Duplicate records:***
We can see from above that there are no exact duplicate records however there are roughly 55,000 duplicate track_ids.  Looking at the data, we can see that the reason why there are duplicate track ids is because there are songs with the same name categorized with different genres.  For example, the song "Lucy Mongoosey" is duplicated because it is classified a Dance genre and an Electronic genre.

***B. Before we remove the duplicate songs, let's pick only the features\columns we will keep for training our nueral network models and call this new dataframe "df_Final". From looking at the data, the relevant columns we should keep are the numeric ones, and we should discard all the text columns except for "key", "mode", "time_signature".  Our target column will be popularity.  We will then remove all duplicates from our newly created "df_Final" dataframe and use this dataframe to train our neural network model using x_features and y_target dataframes.***

```python
numeric_features = ["acousticness", "danceability", "duration_ms", "energy",
                    "instrumentalness", "liveness", "loudness", 
                    "speechiness", "tempo", "valence","popularity"]
# we will keep these string columns and one hot encode them to numeric
categorical_features = ["key", "mode", "time_signature"]
```

```python
#Combine numeric and the few string categories we will keep, then Removed duplicates and save all results to df_temp dataframe
df_temp = df_spotify[numeric_features + categorical_features]
df_temp.duplicated().sum()
#let's remove duplicates
df_temp=df_temp.drop_duplicates()
df_temp.duplicated().sum()
```

```python

# Preprocess (scale numeric + one-hot categorical)
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_features),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
])

df_Final = preprocessor.fit_transform(df_temp)

# Get feature names (numeric + encoded categorical)
encoded_cat = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)
all_features = numeric_features + list(encoded_cat)

# Convert back to DataFrame
df_Final = pd.DataFrame(df_Final, columns=all_features)
y_target = df_Final["popularity"]
x_features = df_Final.drop(["popularity"], axis=1)
y_target.head()
```

# ***5. Create Neural Network Model***

A. First let's split our data into training and validation

```python
X_train, X_test, y_train, y_test = train_test_split(x_features, y_target, test_size=0.2, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x_features, y_target, test_size=0.2, random_state=42)

# Convert to tensors
X_train_tensor = torch.tensor(X_train.values, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).view(-1, 1)
X_test_tensor = torch.tensor(X_test.values, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.float32).view(-1, 1)

# DataLoader
train_loader = DataLoader(TensorDataset(X_train_tensor, y_train_tensor), batch_size=32, shuffle=True)
test_loader = DataLoader(TensorDataset(X_test_tensor, y_test_tensor), batch_size=32)
```

```python
class SongPopularityRegressor(nn.Module):
    def __init__(self, input_size):
        super(SongPopularityRegressor, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_size, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(128, 64),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(64, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.Linear(32, 1)   # 207.68 No sigmoid! Output is raw popularity score
        )
        '''self.net = nn.Sequential(
            nn.Linear(input_size, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(16, 1)  # 207.68 No sigmoid! Output is raw popularity score
        )'''

    def forward(self, x):
        return self.net(x)

model = SongPopularityRegressor(input_size=X_train.shape[1])
```

```python
#loss function
criterion = nn.MSELoss()  # Mean Squared Error for regression
#optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
optimizer = torch.optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)

# Scheduler
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',        # because we want to minimize loss
    factor=0.5,        # reduce LR by half
    patience=5,        # wait 5 epochs with no improvement
    min_lr=1e-6,       # don’t go below this
    verbose=True
)
```

```python
#train model
epochs = 100

'''for epoch in range(epochs):
    model.train()
    total_loss = 0
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"Epoch {epoch+1}/{epochs}, Loss: {total_loss/len(train_loader):.4f}")'''
    
    
for epoch in range(epochs):
    # Training phase
    model.train()
    total_loss = 0
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    
    # Validation phase
    model.eval()
    val_loss = 0
    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            val_loss += loss.item()
    
    # Print results per epoch
    print(
        f"Epoch {epoch+1}/{epochs}, "
        f"Train Loss: {total_loss/len(train_loader):.4f}, "
        f"Val Loss: {val_loss/len(test_loader):.4f}"
    )
    
    
    
```

```python
#evaluate model
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

model.eval()
with torch.no_grad():
    y_pred = model(X_test_tensor).numpy().flatten()
    y_true = y_test_tensor.numpy().flatten()

mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_true, y_pred)
r2 = r2_score(y_true, y_pred)

print(f"MSE: {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"R²: {r2:.2f}")
```
